# **Final summary of all datasets**

This notebook compares the performance of different machine learning models across four datasets: Iris , Titanic, Diabetes, Banana .
The models are evaluated using accuracy scores.

```python
import pandas as pd
import matplotlib as plt

# Datasets
datasets= ["Iris", "Titanic", "Diabetes","Banana "]

# Accuracy scores
results ={ "Naive Bayes":[ 0.96 , 0.78, 0.76, 0.90],
          "Neural Network": [ 0.98, 0.82, 0.80 , 0.94],
           "Decision Tree " : [ 0.97, 0.79, 0.75, 0.91],
           "Random Forest" : [ 0.99, 0.85, 0.82, 0.96],
           "SVM" :[ 0.97, 0.83, 0.78, 0.93],
           "KNN":[ 0.96, 0.80, 0.77, 0.92]
           }
# Create Table
df = pd.DataFrame(results , index= datasets)

df.plot(kind='bar')
plt.title("Model Camparison Across Datasets")
plt.ylabel("Accuracy")
plt.xlabel("Datasets")
plt.show()
print("Fianl Accuracy Camparison Across Datasets ")
print(df)
```

# Observation :-
Random Forest performed best across most datasets.
Titanic and Diabetes datasets showed lower accuracy compared to Iris and Banana.
